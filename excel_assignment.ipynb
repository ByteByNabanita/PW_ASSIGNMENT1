Ans 1: Difference between relative, absolute and mixed cell referencing-

A relative reference changes when we copy the formula. Both row and column shifts to match new position.
For example, =B2 * C2 in E2 becomes =B3 * C3 in E3. It is used when we want the reference to adjust automatically.

An absolute reference stays fixed when we copy the formula. Row and column stay locked, never change. For example, $E$1 means both column and row is fixed and does not shift to D3,D4 etc when copied down. It is used when one value is used repeatedly.

A mixed cell reference fixes either the row or the column, but not both.  $A1 means column A remains fixed while the row changes.  A$1 means row 1 stays fixed while the column changes.
